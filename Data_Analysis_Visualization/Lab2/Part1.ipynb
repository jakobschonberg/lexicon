```python
import pandas as pd
listings = pd.read_csv('../data/listings.csv')
```

# 1)

```python
listings['property_type'].value_counts(normalize=True).head().round(3)
```

# 2)

```python
listings['neighbourhood_cleansed'].value_counts(ascending=True).head(3)
```

# 3)

```python
listings.groupby('room_type')['beds'].agg(
    ['count', 'mean']
).round(3)
```

All averages are accurate. What do you mean trust the least?
If you are asking which average says the most, obviously the ones with more data are more statistically significant ('Entire home/apt' and 'Private room') and less likely to be skewed by a few odd data points.

# 4)

```python
listings.groupby('neighbourhood_cleansed')['number_of_reviews'].agg(
    ['count', 'mean', 'median']).sort_values('median').round(2)
```

All neighbourhoods have much higher mean than median, likely because there's a large number of listings with 0 or just a few reviews and a few listings that have a lot of reviews to drive the average up.

# 5)

```python
listings.groupby('room_type').agg(
    n_listings=('id', 'count'),
    avg_beds=('beds', 'mean'),
    max_reviews=('number_of_reviews', 'max')
).round(2)
```

# 6)

```python
(pd.crosstab(listings['neighbourhood_cleansed'], listings['host_is_superhost'],
             normalize='index')*100).sort_values(by='t').round(1)
```

# 7a)

```python
listings['has_reviews'] = listings['number_of_reviews'] > 0
print((listings['has_reviews']==False).sum())
```

# b)

```python
import numpy as np
listings['popularity'] = np.where(
    listings['number_of_reviews'] >= 50, 'popular', 'other'
)
listings['popularity'].value_counts()
```

# 8)

```python
result = pd.cut(listings['minimum_nights'],
       bins=[1,6,29,1500000],
       labels=['2-6', '7-29','30+']
)
print(result.isna().sum())
result.value_counts().sort_index()
```

```python
result = pd.cut(listings['minimum_nights'],
       bins=[-1000,1,6,29,1500000],
       labels=['1', '2-6', '7-29','30+']
)
print(result.isna().sum())
result.value_counts().sort_index()
```
